```polyglot-notebook
using System;
using System.Collections.Generic;
using System.Text;

public abstract class Component
{
    public string Model { get; set; }
    public decimal Price { get; set; }

    public Component(string model, decimal price)
    {
        Model = model;
        Price = price;
    }

    public virtual string GetInfo()
    {
        return $"Компонент: {Model}, Цена: {Price}";
    }

    // Перегрузка ToString()
    public override string ToString()
    {
        return GetInfo();
    }
}

public interface IInstallable
{
    void Install();
}


public class CPU : Component, IInstallable
{
    public int Cores { get; set; }

    public CPU(string model, decimal price, int cores) : base(model, price)
    {
        Cores = cores;
    }

    public override string GetInfo()
    {
        return $"CPU: {Model}, Ядро: {Cores}, Цена: {Price}";
    }

    public void Install()
    {
        Console.WriteLine($"Устанавливаем CPU {Model}...");
    }

    public static bool operator ==(CPU a, CPU b)
    {
        if (ReferenceEquals(a, b)) return true;
        if (a is null || b is null) return false;
        return a.Model == b.Model && a.Cores == b.Cores && a.Price == b.Price;
    }

    public static bool operator !=(CPU a, CPU b)
    {
        return !(a == b);
    }

    public override bool Equals(object obj)
    {
        return obj is CPU cpu && this == cpu;
    }

    public override int GetHashCode()
    {
        return HashCode.Combine(Model, Cores, Price);
    }
}

public class RAM : Component, IInstallable
{
    public int CapacityGB { get; set; }

    public RAM(string model, decimal price, int capacityGB) : base(model, price)
    {
        CapacityGB = capacityGB;
    }

    public override string GetInfo()
    {
        return $"RAM: {Model}, Емкость: {CapacityGB}Гб, Цена: {Price}";
    }

    public void Install()
    {
        Console.WriteLine($"Устанавливаем RAM {Model}...");
    }

    public static bool operator ==(RAM a, RAM b)
    {
        if (ReferenceEquals(a, b)) return true;
        if (a is null || b is null) return false;
        return a.Model == b.Model && a.CapacityGB == b.CapacityGB && a.Price == b.Price;
    }

    public static bool operator !=(RAM a, RAM b)
    {
        return !(a == b);
    }

    public override bool Equals(object obj)
    {
        return obj is RAM ram && this == ram;
    }

    public override int GetHashCode()
    {
        return HashCode.Combine(Model, CapacityGB, Price);
    }
}

public class GPU : Component, IInstallable
{
    public int MemoryGB { get; set; }

    public GPU(string model, decimal price, int memoryGB) : base(model, price)
    {
        MemoryGB = memoryGB;
    }

    public override string GetInfo()
    {
        return $"GPU: {Model}, Емкость: {MemoryGB}GB, Цена: {Price:C2}";
    }

    public void Install()
    {
        Console.WriteLine($"Устанавливаем GPU {Model}...");
    }

    public static bool operator ==(GPU a, GPU b)
    {
        if (ReferenceEquals(a, b)) return true;
        if (a is null || b is null) return false;
        return a.Model == b.Model && a.MemoryGB == b.MemoryGB && a.Price == b.Price;
    }

    public static bool operator !=(GPU a, GPU b)
    {
        return !(a == b);
    }

    public override bool Equals(object obj)
    {
        return obj is GPU gpu && this == gpu;
    }

    public override int GetHashCode()
    {
        return HashCode.Combine(Model, MemoryGB, Price);
    }
}

public class HDD : Component, IInstallable
{
    public int CapacityGB { get; set; }

    public HDD(string model, decimal price, int capacityGB) : base(model, price)
    {
        CapacityGB = capacityGB;
    }

    public override string GetInfo()
    {
        return $"HDD: {Model}, Емкость: {CapacityGB}GB, Цена: {Price:C2}";
    }

    public void Install()
    {
        Console.WriteLine($"Устанавливаем HDD {Model}...");
    }

    public static bool operator ==(HDD a, HDD b)
    {
        if (ReferenceEquals(a, b)) return true;
        if (a is null || b is null) return false;
        return a.Model == b.Model && a.CapacityGB == b.CapacityGB && a.Price == b.Price;
    }

    public static bool operator !=(HDD a, HDD b)
    {
        return !(a == b);
    }

    public override bool Equals(object obj)
    {
        return obj is HDD hdd && this == hdd;
    }

    public override int GetHashCode()
    {
        return HashCode.Combine(Model, CapacityGB, Price);
    }
}

// Класс Computer с Generic методами
public class Computer
{
    private List<Component> components = new List<Component>();

    public void AddComponent<T>(T component) where T : Component, IInstallable
    {
        components.Add(component);
        component.Install();
    }

    public void RemoveComponent<T>(T component) where T : Component
    {
        components.Remove(component);
    }

    public decimal GetTotalPrice()
    {
        decimal total = 0m;
        foreach (var c in components)
        {
            total += c.Price;
        }
        return total;
    }

    public void ShowAllComponents()
    {
        Console.WriteLine("Компоненты компьютера:");
        foreach (var c in components)
        {
            Console.WriteLine(c.ToString());
        }
    }
}

// Статический класс с методами создания конфигураций
public static class ComputerFactory
{
    public static Computer CreateGamingPC()
    {
        Computer computer = new Computer();
        computer.AddComponent(new CPU("CPU", 600m, 24));
        computer.AddComponent(new RAM("RAM 32GB", 150m, 32));
        computer.AddComponent(new GPU("NVIDIA RTX 4090", 1600m, 24));
        computer.AddComponent(new HDD("HDD 1TB ", 130m, 1000));
        return computer;
    }

    public static Computer CreateOfficePC()
    {
        Computer computer = new Computer();
        computer.AddComponent(new CPU("Intel i5-12400", 200m, 6));
        computer.AddComponent(new RAM("RAM 16GB", 70m, 16));
        computer.AddComponent(new GPU("GPU", 0m, 0));
        computer.AddComponent(new HDD("HDD 500GB", 50m, 500));
        return computer;
    }
}

var homePC = ComputerFactory.CreateGamingPC();
homePC.ShowAllComponents();
Console.WriteLine($"Общая цена: {homePC.GetTotalPrice()}\n");

var officePC = ComputerFactory.CreateOfficePC();
officePC.ShowAllComponents();
Console.WriteLine($"Общая цена: {officePC.GetTotalPrice()}\n");

GPU gpuToRemove = new GPU("GPU", 0m, 0);
officePC.RemoveComponent(gpuToRemove);
Console.WriteLine("После удаления GPU:");
officePC.ShowAllComponents();
Console.WriteLine($"Общая цена: {officePC.GetTotalPrice()}\n");

var cpu1 = new CPU("CPU1", 200m, 6);
var cpu3 = new CPU("CPU3", 600m, 24);
Console.WriteLine($"CPU1 == CPU3: {cpu1 == cpu3}");
Console.WriteLine(cpu1.Equals(cpu3));
```
